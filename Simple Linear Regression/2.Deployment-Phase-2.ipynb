# Model Deployment — Inference Pipeline

## Deployment Objective
This notebook demonstrates how to load the trained Linear Regression model and generate salary predictions for unseen input data.

## Deployment Steps
1. Load the serialized model
2. Prepare inference input
3. Execute prediction
4. Display the predicted salary

```python
# Import pickle library to load the existing model.
import pickle
import pandas as pd
```

```python
# Load the existing model and pass the input as 'zero' to calculate the salary of the fresher.
loaded_model=pickle.load(open("finalized_model_linear.sav",'rb'))
feature_names = ['YearsExperience']
```

```python
# Get the input from the user for the years of experience.
while True:
    try:        
        years_of_experience = int(input("Enter the years of experience for which the salary to be predicted: "))
        if years_of_experience < 0:
            raise ValueError
        break
    except ValueError:
        print("Entered number is not valid integer. Enter an integer number")
```

```python
# Assign the user input to the model to predict the salary.
new_data = pd.DataFrame([[years_of_experience]], columns=feature_names)
result=loaded_model.predict(new_data)
```

```python
# Print the predicted salary.
print(f"The salary for the {years_of_experience} years of experience: {result}")
```
