# Model Deployment — Inference Pipeline

## Deployment Objective
This notebook demonstrates how to load the trained Linear Regression model and generate salary predictions for unseen input data.

## Deployment Steps
1. Load the serialized model
2. Prepare inference input
3. Execute prediction
4. Display the predicted salary

```python
# Import pickle library to load the existing model.
import pickle
import pandas as pd
```

```python
# Load the existing model and pass the input as 'zero' to calculate the salary of the fresher.
loaded_model=pickle.load(open("final_mulit_linear_model.sav",'rb'))
feature_names = ['R&D Spend', 'Administration', 'Marketing Spend', 'State_Florida', 'State_New York']
```

```python
# Assign the user input to the model to predict the salary.
new_data = pd.DataFrame([[1234,345,4565,1,0]], columns=feature_names)
result=loaded_model.predict(new_data)
```

```python
# Print the profit if a startup has started at Florida.
print(f"The profit if a startup has started at Florida: {result}")
```
